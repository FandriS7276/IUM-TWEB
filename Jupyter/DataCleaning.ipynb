# Data Cleaning

### Import of pandas

```python
import pandas as pd
```

## Actors file cleaning

```python
# Showing number of rows in uncleaned dataset
actors_cleaned = pd.read_csv('../Csv/actors.csv')
print("Numbers of rows in the actors dataset: ", len(actors_cleaned))
```

### Removing any empty rows and rows with selectioned null field

```python
# Remove rows with empty 'id'
actors_cleaned = actors_cleaned.dropna(subset=['id'])
print("Cleaning any empty id: ",len(actors_cleaned))

# Remove rows with empty 'name'
actors_cleaned = actors_cleaned.dropna(subset=['name'])
print("Cleaning any empty name: ",len(actors_cleaned))
```

### Removing duplicate lines

```python
# Remove duplicate rows
actors_cleaned = actors_cleaned.drop_duplicates(subset=['id','name','role'], keep='first')
print("Numbers of lines in actors_cleaned (no duplicates): ", len(actors_cleaned), "\n")

# Sort the DataFrame for better readability
actors_cleaned = actors_cleaned.sort_values(by=['id','name'])

# Print final display and save to new cleaned csv
print("Displaying the cleaned data frame:")
print(actors_cleaned.head().to_string(index=False, col_space=10))
actors_cleaned.to_csv('../Csv/actors_cleaned.csv', index=False)
```

## Movies file cleaning

```python
# Showing number of rows in uncleaned dataset
movies_cleaned = pd.read_csv('../Csv/movies.csv')
print("Numbers of rows in the movies dataset: ", len(movies_cleaned))
```

### Removing any empty rows and rows with selectioned null field

```python
# Remove rows with empty 'id'
movies_cleaned = movies_cleaned.dropna(subset=['id'])
print("Cleaning any empty id: ",len(movies_cleaned))

# Remove rows with empty 'name'
movies_cleaned = movies_cleaned.dropna(subset=['name'])
print("Cleaning any empty name: ",len(movies_cleaned))

# Remove rows with empty 'date'
movies_cleaned = movies_cleaned.dropna(subset=['date'])
# Solve date turning into float issue
movies_cleaned['date'] = movies_cleaned['date'].astype(int)
print("Cleaning any empty date: ",len(movies_cleaned))
```

### Removing duplicate lines

```python
# Remove duplicate rows
movies_cleaned = movies_cleaned.drop_duplicates(
    subset=['id','name','date','tagline','description','minute','rating'], keep='first')
print("Numbers of lines in movies_cleaned (no duplicates): ", len(movies_cleaned), "\n")

# Solve date turning into float issue
movies_cleaned['minute'] = movies_cleaned['minute'].astype('Int64')

# Print final display and save to new cleaned csv
print("Displaying the cleaned data frame:")
print(movies_cleaned.head().to_string(index=False, col_space=10))
movies_cleaned.to_csv('../Csv/movies_cleaned.csv', index=False)
```
