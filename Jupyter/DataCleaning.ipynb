# Data Cleaning
### In this executable we are gonna reduce the amount of errors and missing data that could otherwise hinder our analysis by removing the rows that are either duplicates or have missing primary keys which are essential to identify each record and ensure data integrity.

## Import Pandas

```python
import pandas as pd
```

## Data collection

```python
actors = pd.read_csv('../datasets/actors.csv')
countries = pd.read_csv('../datasets/countries.csv')
crew = pd.read_csv('../datasets/crew.csv')
genres = pd.read_csv('../datasets/genres.csv')
languages = pd.read_csv('../datasets/languages.csv')
movies = pd.read_csv('../datasets/movies.csv')
posters = pd.read_csv('../datasets/posters.csv')
releases = pd.read_csv('../datasets/releases.csv')
rotten_tomatoes_reviews = pd.read_csv('../datasets/rotten_tomatoes_reviews.csv')
studios = pd.read_csv('../datasets/studios.csv')
the_oscar_awards = pd.read_csv('../datasets/the_oscar_awards.csv')
themes = pd.read_csv('../datasets/themes.csv')
```

## Data showcase settings

```python
# Function to print dataframe without index
def printni(df, rows=10):
    print(df.head(rows).to_string(index=False) + '\n')

# Function to print number of rows in dataframe
def printr(df):
    name = [n for n, value in globals().items() if value is df][0]
    print(f"Number of rows in {name}: {len(df)}\n")

# Function to inspect dataframe: print number of rows and first few rows
def inspect(df, rows=10, colwidth=None):
    
    # Automatically get the variable name of the dataframe
    name = [n for n, value in globals().items() if value is df][0]
    
    print(f"Number of rows in {name}: {len(df)}\n")
    
    # Shortens the maximum length of the values in the dataframe to avoid truncation
    
    if colwidth is not None:
        
        # Create a copy of the head of the dataframe to avoid modifying the original dataframe
        head_df = df.head(rows).copy()
        
        # Shortens the values in the dataframe to the specified column width, adding '..'
        # if the value is longer than the column width
        shorten = lambda x: (str(x)[:colwidth] + '..') if len(str(x)) > colwidth else str(x)
        
        # Apply the shortening function to all values in the dataframe
        head_df = head_df.map(shorten)
        print(head_df.to_string(index=False, col_space=30)+ '\n')
        
    else:
        print(df.head(rows).to_string(index=False, col_space=30))
```

## Actors file cleaning
### Dataset showcase

```python
# Showing number of rows in uncleaned dataset
inspect(actors)
```

### Duplicates and NULL columns removal

```python
# Removing rows that have missing values in either "id" or "name" columns
# This happens because dropna uses by default how="any", which means that if any of the specified columns
# has a missing value, the entire row will be dropped.
actors_cleaned = actors.dropna(subset=['id', 'name'])
print('NULL removal:')
printr(actors_cleaned)

# Removing duplicates
actors_cleaned = actors_cleaned.drop_duplicates(subset=['id','name','role'], keep='first')
print('Duplicates removal:')
printr(actors_cleaned)
```

### Sort, final showcase and exporting dataframe to CSV

```python
# Sort the DataFrame for better readability
actors_cleaned = actors_cleaned.sort_values(by=['id','name'])

# Print final display and save to new cleaned csv
print("Displaying the cleaned data frame:")
printni(actors_cleaned)
actors_cleaned.to_csv('../datasets/actors_cleaned.csv', index=False)
```
